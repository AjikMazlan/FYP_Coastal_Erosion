```python
physical_numeric = pd.to_numeric(
    df_raw["Physical_Erosion_Score"],
    errors="coerce"
)

economic_numeric = pd.to_numeric(
    df_raw["Economic_Score"],
    errors="coerce"
)

calculated_total = physical_numeric * economic_numeric

score_check = df_raw[
    physical_numeric.notna()
    & (
        abs(calculated_total - df_raw["Total_Score"]) > 0.01
    )
][
    [
        "No",
        "Coastal_Cell",
        "Area_Location",
        "Physical_Erosion_Score",
        "Economic_Score",
        "Total_Score"
    ]
]

score_check
```

```python
expected_numbers = set(
    range(
        int(df_raw["No"].min()),
        int(df_raw["No"].max()) + 1
    )
)

actual_numbers = set(df_raw["No"])

missing_station_numbers = sorted(
    expected_numbers - actual_numbers
)

print("Missing station numbers:")
print(missing_station_numbers)

print("Number of missing station numbers:",
      len(missing_station_numbers))
```

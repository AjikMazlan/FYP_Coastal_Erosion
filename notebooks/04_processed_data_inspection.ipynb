# 04 - Inspect the current processed coastal dataset

Read only the processed `_144.csv` dataset produced by Notebook 03. Distinguish
original JPS observations from interpolated points, and verify that missing
erosion measurements are confined to Protected stations. The current cleaned
score column is `Total_Score_Cleaned`.


```python
import pandas as pd
import numpy as np

from pathlib import Path

# Locate the project from either the repository root or its notebooks folder.
PROJECT_ROOT = Path.cwd().resolve()
while not (PROJECT_ROOT / "data" / "raw" / "jps_coastal_erosion_full.csv").is_file():
    if PROJECT_ROOT.parent == PROJECT_ROOT:
        raise FileNotFoundError("Open this notebook from inside FYP_Coastal_Erosion.")
    PROJECT_ROOT = PROJECT_ROOT.parent

df = pd.read_csv(
    PROJECT_ROOT / "data/processed/terengganu_coastal_erosion_transects_144.csv"
)
```

```python
print("Dataset shape:", df.shape)

print("\nSource distribution:")
print(df["Source_Type"].value_counts())

print("\nProtected points:", df["Is_Protected"].sum())

print("\nDataset preview:")
display(df.head())
```

```python
df.info()
```

```python
print(df.isnull().sum())
```

```python
missing_erosion = df[
    df["Annual_Erosion_Rate_m_yr"].isna()
]

print("Rows with missing erosion:", len(missing_erosion))

print(
    "\nProtection status of missing erosion rows:"
)

print(
    missing_erosion["Is_Protected"].value_counts()
)
```

```python
print(
    "Duplicate rows:",
    df.duplicated().sum()
)

print(
    "Duplicate Transect IDs:",
    df["Transect_ID"].duplicated().sum()
)

print(
    "Duplicate coordinates:",
    df.duplicated(
        subset=["Latitude_DD", "Longitude_DD"]
    ).sum()
)
```

```python
print(
    "Latitude:",
    df["Latitude_DD"].min(),
    "to",
    df["Latitude_DD"].max()
)

print(
    "Longitude:",
    df["Longitude_DD"].min(),
    "to",
    df["Longitude_DD"].max()
)
```

```python
df[
    [
        "Annual_Erosion_Rate_m_yr",
        "Physical_Erosion_Score",
        "Economic_Score",
        "Total_Score_Cleaned"
    ]
].describe()
```

```python
df["Coastal_Cell"].value_counts().sort_index()

print(
    "Number of coastal cells:",
    df["Coastal_Cell"].nunique()
)

print(
    "Number of districts:",
    df["District"].nunique()
)

df["District"].value_counts()

protection_counts = df["Is_Protected"].value_counts()

print(protection_counts)

print(
    "\nProtected percentage:",
    round(
        df["Is_Protected"].mean() * 100,
        2
    ),
    "%"
)
```

```python
print("=== FINAL PROCESSED DATASET CHECK ===")

print("Rows:", len(df))
print("Columns:", len(df.columns))

print(
    "Unique Transect IDs:",
    df["Transect_ID"].nunique()
)

print(
    "Coastal cells:",
    df["Coastal_Cell"].nunique()
)

print(
    "Districts:",
    df["District"].nunique()
)

print(
    "Protected points:",
    df["Is_Protected"].sum()
)

print(
    "Duplicate rows:",
    df.duplicated().sum()
)

features = ["Annual_Erosion_Rate_m_yr", "Physical_Erosion_Score", "Economic_Score"]
original_mask = df["Source_Type"].eq("JPS_Original")
protected_mask = df["Is_Protected"].eq(1)
assert df["Transect_ID"].is_unique
assert df[["Latitude_DD", "Longitude_DD"]].notna().all().all()
assert original_mask.sum() == 144
assert df.loc[original_mask, "No"].nunique() == 144
assert df.loc[~protected_mask, features].notna().all().all()
assert df.loc[protected_mask, "Annual_Erosion_Rate_m_yr"].isna().all()
print("All processed dataset checks passed.")
```

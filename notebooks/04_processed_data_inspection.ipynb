```python
# Step 2: Check dataset quality

features = [
    "Annual_Erosion_Rate_m_yr",
    "Physical_Erosion_Score",
    "Economic_Score"
]

print("=== DATA QUALITY CHECK ===")

print("Total points:", len(df))

print("\nDuplicate Transect IDs:")
print(df["Transect_ID"].duplicated().sum())

print("\nMissing coordinates:")
print(df[
    ["Latitude_DD", "Longitude_DD"]
].isnull().sum())

print("\nMissing AI features (non-protected only):")
print(
    df.loc[df["Is_Protected"] == 0, features]
    .isnull().sum()
)

print("\nOriginal JPS station numbers:")
print(
    df.loc[df["Source_Type"] == "JPS_Original", "No"]
    .nunique()
)
```

```python
# Step 3: Check erosion data ranges

df_valid = df[df["Is_Protected"] == 0]

print("=== EROSION DATA SUMMARY ===")

print(df_valid[
    ["Annual_Erosion_Rate_m_yr",
     "Physical_Erosion_Score",
     "Economic_Score"]
].describe().round(2))

print("\nNegative erosion rates:",
      (df_valid["Annual_Erosion_Rate_m_yr"] < 0).sum())

print("\nMaximum erosion rate:",
      df_valid["Annual_Erosion_Rate_m_yr"].max())

print("\nNon-protected points:", len(df_valid))
print("Protected points:", df["Is_Protected"].sum())
```
